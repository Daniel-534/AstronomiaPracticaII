```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("asu.tsv", sep=";", comment="#")

df = df.iloc[2:].reset_index(drop=True)
```

```python
x = pd.to_numeric(df["BP-RP"].str.strip(), errors='coerce')
y = pd.to_numeric(df["Gmag"].str.strip(), errors='coerce')
```

```python
plt.figure(figsize=(8, 6))
plt.scatter(x, y, s=15, color='black', alpha=0.7)
plt.gca().invert_yaxis()  # Magnitudes invertidas en diagramas HR 
plt.xlabel("BP - RP (IndiceDeColor)")
plt.ylabel("Gmag (MagnitudAparente)")
plt.title("Diagrama HR - Gaia DR3")
plt.grid(True, linestyle="--", alpha=0.5)
```
